```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow

rmin = 100
rmax = 200

img = cv2.imread("/content/goat.jpg", 0)

slice_with_bg = img.copy()
slice_without_bg = np.zeros(img.shape, np.uint8)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):

        pixel_value = img[i][j]

        if rmin <= pixel_value <= rmax:
            slice_with_bg[i][j] = 255
            slice_without_bg[i][j] = 255
        else:
            slice_with_bg[i][j] = pixel_value
            slice_without_bg[i][j] = 0
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap='gray')
plt.title("Original")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(slice_with_bg, cmap='gray')
plt.title("Slice with Background")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(slice_without_bg, cmap='gray')
plt.title("Slice without Background")
plt.axis("off")

plt.show()
```
